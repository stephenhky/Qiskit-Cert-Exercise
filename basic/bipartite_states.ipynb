```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
epr_circuit = QuantumCircuit(2)
epr_circuit.h(0)
epr_circuit.cx(0, 1)
```

```python
epr_circuit.draw()
```

```python
# ev = Statevector.from_int(0, 4)
ev = Statevector.from_label("00")
ev_vec1 = ev.evolve(epr_circuit)
```

```python
ev_vec1.draw("latex")
```

```python
ev_vec1.draw("qsphere")
```

```python
ev_vec1.draw("hinton")
```

```python
singlet_circuit = QuantumCircuit(2)
singlet_circuit.h(0)
singlet_circuit.x(1)
singlet_circuit.cx(0, 1)
```

```python
singlet_circuit.draw()
```

```python
ev_vec2 = ev.evolve(singlet_circuit)
```

```python
ev_vec2.draw("latex")
```

```python
ev_vec2.draw("qsphere")
```

```python
ev_vec2.draw("hinton")
```
