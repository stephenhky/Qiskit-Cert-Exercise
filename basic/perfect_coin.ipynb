```python
from qiskit.circuit import Gate, QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit import transpile
from qiskit_aer import StatevectorSimulator
```

```python
def perfect_coin_gate() -> Gate:
    qr = QuantumRegister(1)
    qc = QuantumCircuit(qr)

    qc.h(qr[0])

    return qc.to_gate()
```

```python
qc = QuantumCircuit(1, 1)
qc.append(perfect_coin_gate(), [0])
qc.measure(0, 0)
```

```python
transpiled_qc = transpile(qc)
```

```python
simulator = StatevectorSimulator(max_shot_size=1024)
```

```python
simulation_results = simulator.run(transpiled_qc)
```

```python
simulation_results.result()
```
