```python
from qiskit.circuit import Gate, QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit import transpile
from qiskit_aer import StatevectorSimulator
```

```python
def perfect_coin_gate() -> Gate:
    qr = QuantumRegister(1)
    qc = QuantumCircuit(qr)
    qc.h(qr[0])
    return qc.to_gate()
```

```python
qr = QuantumRegister(1)
cr = ClassicalRegister(1)
qc = QuantumCircuit(qr, cr)
qc.append(perfect_coin_gate(), [qr[0]])
qc.measure(qr[0], cr[0])
```

```python
simulator = StatevectorSimulator()
```

```python
transpiled_qc = transpile(qc, simulator, optimization_level=0)
```

```python
job = simulator.run(transpiled_qc, shots=1024)
```

```python
count_dict = job.result().get_counts(qc)
```
