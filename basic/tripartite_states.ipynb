```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
ghz_circuit = QuantumCircuit(3)
ghz_circuit.h(0)
ghz_circuit.cx([0, 0], [1, 2])
```

```python
ghz_circuit.draw()
```

```python
ghz_circuit.draw("mpl")
```

```python
# ev = Statevector.from_label("000")
ev = Statevector.from_int(0, 2**3)
```

```python
ev_vec1 = ev.evolve(ghz_circuit)
```

```python
ev_vec1.draw("latex")
```

```python
ev_vec1.draw("qsphere")
```

```python
ev_vec1.draw("hinton")
```
