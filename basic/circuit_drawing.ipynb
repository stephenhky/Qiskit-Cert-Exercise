```python
from qiskit import QuantumCircuit, QuantumRegister
```

```python
q = QuantumRegister(1)
qc = QuantumCircuit(q)
qc.x(q[0])
qc.h(q[0])
style = {'backgroundcolor': 'lightgreen'}
qc.draw(output='mpl', style=style, scale=3.5, plot_barriers=False, reverse_bits=False)
```

```python
qc.inverse().draw("mpl")
```
