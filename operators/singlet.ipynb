```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, DensityMatrix, Operator
```

```python
singlet_qc = QuantumCircuit(2)
singlet_qc.h(0)
singlet_qc.x(1)
singlet_qc.cx(0, 1)
```

```python
singlet_qc.draw("mpl")
```

```python
ev = Statevector.from_label("00")
```

```python
ev1 = ev.evolve(singlet_qc)
```

```python
ev1.draw("latex")
```

```python
denmat = DensityMatrix.from_label("00")
```

```python
denmat
```

```python
denmat.evolve(singlet_qc)
```

```python
singlet_qc_operator = Operator(singlet_qc)
```

```python
singlet_qc_operator
```

```python
denmat.evolve(singlet_qc_operator)
```
