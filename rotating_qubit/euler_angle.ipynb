```python
from math import pi

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
ev = Statevector.from_int(0, dims=2)
```

```python
qc1 = QuantumCircuit(1)
qc1.ry(7*pi/4, 0)
```

```python
qc1.draw("mpl")
```

```python
ev1 = ev.evolve(qc1)
```

```python
ev1.draw("latex")
```

```python
ev1.draw("bloch")
```
