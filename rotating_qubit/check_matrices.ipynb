```python
from math import pi

from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator, Pauli
```

```python
qc_x = QuantumCircuit(1)
qc_x.rx(pi / 4., 0)
```

```python
Operator(qc_x)
```

```python
qc_y = QuantumCircuit(1)
qc_y.ry(pi / 4., 0)
```

```python
Operator(qc_y)
```

```python
qc_z = QuantumCircuit(1)
qc_z.rz(pi / 4., 0)
```

```python
Operator(qc_z)
```
