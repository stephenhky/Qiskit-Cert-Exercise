```python
from math import pi

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
ev = Statevector.from_int(0, dims=2)
```

```python
qcx = QuantumCircuit(1)
qcx.u(pi/2, 0, 0, 0)
```

```python
qcx.draw("mpl")
```

```python
evx = ev.evolve(qcx)
```

```python
evx.draw("latex")
```

```python
evx.draw("bloch")
```
