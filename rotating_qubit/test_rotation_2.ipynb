```python
from math import pi

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
ev0 = Statevector.from_int(0, dims=2)
```

```python
qc1 = QuantumCircuit(1)
qc1.u(pi/3, 3*pi/4, pi/4, 0)
```

```python
ev1 = ev0.evolve(qc1)
ev1.draw("bloch")
```

```python
qc2 = QuantumCircuit(1)
qc2.u(pi/4, -pi/2, 0, 0)
```

```python
ev2 = ev0.evolve(qc2)
ev2.draw("bloch")
```

```python
qc3 = QuantumCircuit(1)
qc3.rx(pi/2, 0)
```

```python
ev3 = ev0.evolve(qc3)
ev3.draw("bloch")
```

```python
qc4 = QuantumCircuit(1)
qc4.x(0)
qc4.u(5*pi/6, pi/3, 0, 0)
```

```python
ev4 = ev0.evolve(qc4)
ev4.draw("bloch")
```
